```python
from transcriptforge import TranscriptForge, TranscriptCache, clean_transcript, to_paragraphs

forge = TranscriptForge(cache=TranscriptCache('/content/cache'))
VIDEO = 'https://www.youtube.com/watch?v=GJLlxj_dtq8'  # id or any YouTube URL
transcript = clean_transcript(forge.fetch_best(VIDEO, ['hi', 'en']))
for p in to_paragraphs(transcript)[:5]:
    print(f'[{p.timestamp}]', p.text, '\n')
```

```python
# Many videos in parallel, results saved as files, then zipped
from transcriptforge import save_many, make_zip, write_report_csv

ids = [VIDEO]  # add more ids / URLs
report = forge.fetch_many(ids, ['hi', 'en'], workers=3)
print(report.summary())
save_many(report.transcripts, '/content/out', formats=['text', 'srt', 'markdown', 'json'])
write_report_csv(report, '/content/out/_report.csv')
make_zip('/content/out')  # -> /content/out.zip
```
